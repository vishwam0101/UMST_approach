```python
import json

path = "comparison_metrics_umst.json"   # or your maddpg file

with open(path, "r") as f:
    data = json.load(f)

# remove unwanted keys from each delivery record
for d in data["umst_metrics"]["per_delivery_records"]:
    d.pop("times_bundled", None)
    d.pop("num_vehicle_changes", None)

# Save back
with open(path, "w") as f:
    json.dump(data, f, indent=2)

print("Clean-up complete.")
```
